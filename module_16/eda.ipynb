# Data Visualization Using Matplotlib

```python
from matplotlib import pyplot as plt
```

## Line plot

```python
plt.plot(
    [
        "Jan",
        "Feb",
        "Mar",
        "Apr",
        "May",
        "Jun",
        "Jul",
        "Aug",
        "Sep",
        "Oct",
        "Nov",
        "Dec",
    ],
    [150, 172, 195, 210, 188, 245, 290, 310, 265, 230, 215, 290],
    color="blue",
    marker="o",
    linestyle="-",
    label="2025",
)
plt.plot(
    [85, 110, 125, 160, 178, 195, 220, 240],
    color="green",
    marker="*",
    linestyle="-.",
    label="2026",
)
plt.grid(linestyle=':', color='red', alpha=1)
plt.title("Sales Analysis")
plt.legend()
plt.show()
```

## Bar Plot

```python
import matplotlib.pyplot as plt

department_counts = emp["Department"].value_counts()
print(department_counts)

plt.bar(department_counts.index, department_counts.values, width=0.5, edgecolor="black")
plt.title("Department Counts")
plt.xlabel("Departments")
plt.ylabel("Counts")
plt.grid(linestyle="-.", linewidth=1.3, color="purple", alpha=0.6)
plt.show()
```

## Horizontal Bar Plot

```python
import matplotlib.pyplot as plt

gender_counts = placement["gender"].value_counts()
plt.barh(
    gender_counts.index,
    gender_counts.values,
    color="lightblue",
    edgecolor="black",
    height=0.8,
)
plt.xlabel("Count")
plt.ylabel("Gender")
plt.title("Department Bandwidth")
plt.grid(axis="both", linestyle="-", linewidth=0.4, color="darkblue", alpha=0.4)
plt.show()
```

## Histogram

```python
plt.hist(
    placement["age"], bins=10, color="lightblue", edgecolor="skyblue", density=True
)
plt.title("Age Groups")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6.5, 4))
plt.hist(
    x=placement["cgpa"], color="lightblue", edgecolor="skyblue", bins=10, density=True
)
plt.grid(linestyle=":", linewidth=0.6, color="grey", alpha=0.4)
plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")
plt.show()
placement["cgpa"].describe()
```

## Pie

```python
import numpy as np

plt.pie(
    placement["gender"].value_counts().values,
    labels=placement["gender"].value_counts().index,
    colors=["lightblue", "skyblue"],
    autopct="%1.1f%%",
    explode=[0.1, 0],
    shadow=True,
    startangle=90,
)
plt.title("Gender Distribution", color="black")
plt.show()
```

## Subplot

```python
plt.figure(figsize=(12, 6))
plt.subplot(2, 2, 1)
plt.hist(emp["Age"], color="lightblue", edgecolor="blue", bins=10)
plt.title("Age Distribution", weight="bold")
plt.xlabel("Age", weight="bold")
plt.ylabel("Frequency", weight="bold")

plt.subplot(2, 2, 2)
plt.hist(emp["Salary"], color="lightblue", edgecolor="blue", bins=10)
plt.title("Salary Distribution", weight="bold")
plt.xlabel("Salary", weight="bold")
plt.ylabel("Frequency", weight="bold")

plt.subplot(2, 2, 3)
plt.pie(
    x=np.array(emp["Gender"].value_counts().values),
    labels=np.array(emp["Gender"].value_counts().index),
    colors=["coral", "lightgreen"],
    shadow=True,
    autopct="%1.2f%%",
    explode=[0.1, 0],
    startangle=90,
)
plt.title("Gender Distribution", weight="bold")

plt.subplot(2, 2, 4)
plt.bar(emp["Department"].value_counts().index, emp["Department"].value_counts().values)
plt.title("Department Distribution", weight="bold")
plt.xlabel("Departments", weight="bold")
plt.ylabel("Frequency", weight="bold", color="grey")

plt.tight_layout()
plt.show()
```

# Other Plots and when to use

Trend over time	Line Plot	Shows increase/decrease and trends clearly
Sales by Month	Line Plot	Time-series data
Temperature by Day	Line Plot	Continuous progression
Compare departments	Bar Plot	Easy category comparison
Compare genders	Bar Plot	Compare frequencies/counts
Compare average salary by department	Bar Plot	Compare numerical values across categories
Percentage of males vs females	Pie Chart	Shows part-to-whole relationship
Market share of companies	Pie Chart	Shows contribution percentages
Budget allocation	Pie Chart	Shows percentage distribution
Distribution of Age	Histogram	Shows how values are spread
Distribution of Salary	Histogram	Reveals skewness and outliers
Distribution of CGPA	Histogram	Shows concentration of values
Detect outliers in a numeric feature	Box Plot	Outliers are visible immediately
Compare distributions across categories	Box Plot	Compare median, spread, and outliers
Analyze Salary by Department	Box Plot	Compare salary distributions
Understand data distribution shape	Violin Plot	Shows density + distribution
Compare distributions in detail	Violin Plot	More informative than box plots
Relationship between two numerical variables	Scatter Plot	Shows correlation patterns
Age vs Salary	Scatter Plot	Relationship analysis
Experience vs Performance Score	Scatter Plot	Correlation analysis
Correlation between all numeric features	Heatmap	Visual correlation matrix
Find highly correlated features	Heatmap	Easy correlation interpretation
Analyze pairwise relationships	Pair Plot	Multiple scatter plots automatically
Quick EDA of numerical features	Pair Plot	One graph for many relationships
Missing value visualization	Heatmap (Seaborn)	Easily identifies missing data patterns
Frequency of categorical values	Count Plot	Automatically counts categories
Gender distribution	Count Plot	Better than manually creating counts
Department distribution	Count Plot	Quick categorical analysis

Show a trend	Line Plot
Compare categories	Bar Plot
Show percentages	Pie Chart
Understand distribution	Histogram
Detect outliers	Box Plot
Analyze density/distribution	Violin Plot
Analyze relationship between two numeric features	Scatter Plot
Analyze many feature relationships	Pair Plot
Analyze correlations	Heatmap
Count categorical values	Count Plot

## Violinplot

```python
sns.violinplot(
    data=emp,
    x='Salary'
)

plt.show()
```

```python
sns.violinplot(
    data=placement,
    x='cgpa'
)
plt.title("CGPA Distribution", weight='bold')
plt.xlabel("CGPA")
plt.ylabel("Density")
plt.show()
# Where are most students concentrated?
# Is CGPA evenly distributed?
# Are there multiple peaks?
```

```python
sns.violinplot(
    data=emp,
    x='Department',
    y='Salary',
    hue='Gender'
)
plt.show()
# Department
#    Male Distribution
#    Female Distribution
```

```python
sns.violinplot(
    data=titanic,
    y='Age',
    x='Sex'
)
# Which gender has wider age spread?
# Where are most ages concentrated?
# Are there multiple peaks?
```
