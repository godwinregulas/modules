```python
import pandas as pd

data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "gender": [
        "Male",
        "Female",
        "Male",
        "Female",
        "Male",
        "Female",
        "Male",
        "Female",
        "Male",
        "Female",
    ],
    "age": [20, 21, 20, 22, 21, 20, 23, 21, 22, 20],
    "study_hours": [2, 5, 3, 6, 4, 7, 2, 5, 3, 8],
    "attendance": [75, 90, 80, 95, 85, 92, 70, 88, 78, 96],
    "assignments_completed": [6, 9, 7, 10, 8, 10, 5, 9, 7, 10],
    "exam_score": [58, 78, 65, 88, 72, 91, 52, 81, 63, 94],
}

df = pd.DataFrame(data)

df
```

```python
es_g = df.groupby("gender")['exam_score'].mean()
sns.barplot(x=es_g.index, y=es_g.values)
```

```python
# Scatter plot of study_hours vs exam_score using Seaborn

sns.scatterplot(data=df, x="study_hours", y="exam_score")
```

```python
# Calculate the correlation matrix for all numerical columns and visualize it using a Seaborn heatmap
corr_mat = df.corr(numeric_only=True)

sns.heatmap(corr_mat, annot=True)
```

```python
# Calculate the covariance between study_hours and exam_score.

df.cov(numeric_only=True)['study_hours'].drop(['es_g','assignments_completed',"attendance", 'age','student_id'])
```

1. Exam score is dependent to study hours and they are related
2. Female students are scoring morthan Male students in average
