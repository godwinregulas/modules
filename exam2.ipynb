```python
def getGrade(mark):
    if mark >= 90:
        return "A+"
    elif mark >= 80:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 60:
        return "C"
    else:
        return "Fail"

df['Grade']=df['Marks'].apply(getGrade)

# df['Grade']=getGrade(df['Marks'])
df
```
