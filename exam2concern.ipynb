```python
import pandas as pd
```

```python
titanic = pd.read_csv('../Datasets/titanic_dataset.csv')
titanic
```

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(titanic["Sex"].value_counts().index, titanic["Sex"].value_counts().values)
```

```python
titanic['Cabin'] = titanic['Cabin'].fillna(titanic['Cabin'].mode()[0])
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].mean())
titanic
```

```python
plt.figure(figsize=(22,6))
avg = titanic.groupby('Cabin')['Age'].mean()
plt.bar(avg.index, avg.values, edgecolor='black')
plt.xticks(rotation=90)

avg
```

```python
titanic['Deck'] = titanic['Cabin'].astype(str).str[0]
avg_deck = titanic.groupby('Deck')['Age'].mean()
plt.bar(avg_deck.index, avg_deck.values)
avg_deck
```
