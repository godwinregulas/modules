Creating New Features from Existing Data

```python
# Quantity * Price = Total Feature
import pandas as pd
```

```python
df = pd.DataFrame({"Quantity":[2,5],
             "Price":[500,200]})
df
```

```python
df["Total"] = df["Quantity"] * df["Price"]
```

```python
df
```

```python
# BMI
df = pd.DataFrame({"Height":[1.70,1.80],
             "Weight":[65,80]})
df
```

```python
df["BMI"] = df["Weight"] / (df["Height"]**2)
df
```

```python
# Age from Date of Birth
df = pd.DataFrame({"DOB":["2003-06-07", "1999-08-15"]})
df
```

```python
df["DOB"] = pd.to_datetime(df["DOB"])   # previously given text DOB now turned to datetime  --> Timestamp object/dtype 
df
```

```python
current_year = 2026
df["Age"] = current_year - df["DOB"].dt.year     # dt.year accessor is Due to coversion to datetime using to_datetime()
df
```
