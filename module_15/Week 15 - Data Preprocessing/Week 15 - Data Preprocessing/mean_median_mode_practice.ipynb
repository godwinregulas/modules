```python
import pandas as pd
```

```python
df = pd.read_csv("employee_dataset_with_experience.csv")
df
```

```python
df.isnull().sum()
```

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())    # MEAN
df
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())         # MEDIAN
df
```

```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0]) # Mode
df
```

```python
df["Experience"] = df["Experience"].fillna(df["Experience"].mode()[0])
df
```

```python
df["Gender"] = df["Gender"].fillna("Unknown")  # To fill NULL values as unknown   NO change here bcz already filled the NULL values using mode
```

```python
df.loc[ df["Age"] == 30, "Experience"] = 10      # change at a location with condition
df
```

```python
df["Gender"] = df["Gender"].replace("Male","MALE")   # replace value into another value
df
```
