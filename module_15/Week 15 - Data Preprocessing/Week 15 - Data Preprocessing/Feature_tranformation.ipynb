Log Transformation

```python
import pandas as pd
```

```python
df = pd.DataFrame({"Salary":[30000,32000,35000,500000]})
df
```

```python
import numpy as np
df["Log_Salary"] = np.log(df["Salary"])
df
```

POLYNOMIAL FEATURE TRANSFORMATION    -- if values relationship are not linear and have curved relationship

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
```

```python
df = pd.DataFrame({"Area":[2,3,4]})
df
```

```python

poly = PolynomialFeatures(degree=2,include_bias=False) # degree=2,include_bias=False   optional parameters

new_features = poly.fit_transform(df)
print(new_features)
```
