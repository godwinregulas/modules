Embedded Methods of Feature Selection. (Feature Selection during Model Training)

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Salary": [30000, 40000, 50000, 60000, 70000],
    "Height": [170, 175, 180, 185, 190]
})

y = [0, 0, 1, 1, 1]      # Target
```

```python
# Lasso Regression - Feature Selection
from sklearn.linear_model import Lasso

model = Lasso(alpha = 0.1)

model.fit(df,y)

print(model.coef_)
print(model.feature_names_in_)
```

```python
# Ridge Regression - Feature Selection
```

```python
from sklearn.linear_model import Ridge

model = Ridge(alpha = 1)

model.fit(df,y)
print(model.coef_)
```
