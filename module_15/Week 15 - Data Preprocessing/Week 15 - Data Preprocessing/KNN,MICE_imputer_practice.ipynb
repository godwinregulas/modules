```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Age":[22,24,23,None],
    "Experience":[1,2,2,2],
    "Salary":[30000,32000,31000,31500]
})

df
```

```python
from sklearn.impute import KNNImputer
```

```python
imputer = KNNImputer(n_neighbors = 2)       # Take the no of neighbours mean

new_df = imputer.fit_transform(df)
new_df


import numpy as np                               # just to convert the value into 2 values precision
np.set_printoptions(suppress=True,precision=2)  
print(new_df)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({"Age":[22,25,30,None],
                   "Experience":[1,3,8,5],
                   "Salary":[30000,38000,55000,45000]})
df
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
imputer = IterativeImputer()           # iterative imputer also look into another columns and predict using repeatedly imputing Typically regression + random error
                                       # instead of neighbour mean of KNN
new_df = imputer.fit_transform(df)
new_df
```
