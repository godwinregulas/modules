# Polynomial Features

Sometimes Features are not linear\
So We Create new Features like
- X²
- X³\
at different powers



The Model Can learn Curved relationships

```python
df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(include_bias = False)

poly_transformed = poly.fit_transform(df[["study_hours"]])

df[["study_hour","Polynomial_study_hr"]] = pd.DataFrame(poly_transformed, columns=["study_hour", "Polynomial_study_hr"])
df
```

### Label Encoding

```python
# Label Encoding - Each variable get a unique integer (same variable same value)

import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Label_enc_city"] = encoder.fit_transform(df["city"])
df
```

### One Hot Encoding

```python
# One Hot encoding - Create unique Binary Table

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(df[["city"]])

one_hot_encoded = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(["city"]))
one_hot_encoded
```
