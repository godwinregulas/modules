Feature scaling is the process of Scaling the Feature values into a similar scale\
so that no any feature dominate over other features,
because of its larger values.\
and scale downed values are efficiently processed in ML Models with less computations

# Feature Scaling Techniques
- Normalization (MinMax Scaling)
- Standardization (Z-Score Scaling)
- Robust Scaling

Normalization - MinMax Scailing  -- Scale the values into a fixed range of 0 to 1  $$x_{\text{new}}=\frac{x-\min(x)}{\max(x)-\min(x)}$$

# Normalization

```python
import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
numeric_df = df.select_dtypes("number")
```

```python
scaler = MinMaxScaler()
```

```python
scaled_numeric = scaler.fit_transform(numeric_df)
```

```python
scaled_numeric_df = pd.DataFrame(scaled_numeric, columns = numeric_df.columns)
scaled_numeric_df
```

```python
df[scaled_numeric_df.columns] = scaled_numeric_df
df
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

numeric_df = df.select_dtypes("number")

scaled_numeric = scaler.fit_transform(numeric_df)

scaled_numeric_df = pd.DataFrame(scaled_numeric, columns = numeric_df.columns)

df[scaled_numeric_df.columns] = scaled_numeric_df
df
```

# Standardization
Standardization make Mean = 0,  Std Deviation = 1 \
then calculate Z-score of each values and give score based on the distance from the Mean
$$
z = \frac{x - \mu}{\sigma}
$$

# Robust Scaling

If we use Min-Max or Standardization, \
the outlier strongly affects the scaling.

Robust Scaling avoids this by using:

- **Median**
- **IQR (Q3 − Q1)**

The outlier has much less influence because the median is more robust than the mean.

```python
import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})
```

```python
from sklearn.preprocessing import RobustScaler


scaler = RobustScaler()

selected_df = df[["maths","science","attendance"]]

scaled_selected = scaler.fit_transform(selected_df)

scaled_selected_df = pd.DataFrame(scaled_selected, columns = selected_df.columns)

df[scaled_selected_df.columns] = scaled_selected_df
df
```
