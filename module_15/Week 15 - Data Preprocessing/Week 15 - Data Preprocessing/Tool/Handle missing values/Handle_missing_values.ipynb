```python
df["Study_Hours"] = df["Study_Hours"].fillna(df["Study_Hours"].median())
df
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df
```

```python
df["Attendance_%"] = df["Attendance_%"].fillna(df["Attendance_%"].median())
df
```

```python
df["Assignment_Score"] = df["Assignment_Score"].fillna(df["Assignment_Score"].median())
df
```

```python
df["Exam_Score"] = df["Exam_Score"].fillna(df["Exam_Score"].mean())
df
```

# Advanced Imputations 

```python
# KNN - K-Nearest Neighbour
# MICE - Multiple Imputation By Chained Equation
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Age": [20, 25, 21, 19, 30, 22, 20, 28, 21, 23],
    "Study_Hours": [5, 8, None, 3, 2, 6, 4, 9, 5, 7],
    "Attendance": [90, 95, 87, 75, 70, 89, 88, 96, 85, 91],
    "Exam_Score": [82, 91, 84, 68, 65, 86, 79, 94, 81, 88]
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.impute import KNNImputer
```

```python
numeric_df = df.select_dtypes("number")
```

```python
imputer = KNNImputer(n_neighbors=3)   # imputation based on the number of neighbours mean - not order neighbours but most simalar Entries

imputed_numeric_df = imputer.fit_transform(numeric_df)
imputed_numeric_df
```

```python
imputed_numeric_df = pd.DataFrame(imputed_numeric_df, columns=numeric_df.columns)
imputed_numeric_df
```

```python
# assingning to OG
df[imputed_numeric_df.columns] = imputed_numeric_df
df
```

```python
# MICE

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Age": [20, 25, 21, 19, 30, 22, 20, 28, 21, 23],
    "Study_Hours": [5, 8, None, 3, 2, 6, 4, 9, 5, 7],
    "Attendance": [90, 95, 87, 75, 70, 89, 88, 96, 85, 91],
    "Exam_Score": [82, 91, 84, 68, 65, 86, 79, 94, 81, 88]
}

df = pd.DataFrame(data)
```

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
imputer = IterativeImputer()
```

```python
numeric_df = df.select_dtypes("number")
```

```python
imputed_numeric_df = imputer.fit_transform(numeric_df)
```

```python
imputed_numeric_df = pd.DataFrame(imputed_numeric_df, columns = numeric_df.columns)
```

```python
df[imputed_numeric_df.columns] = imputed_numeric_df
df
```
