# Z-score
Z-Score Measures How many standard Deviation a Value is Away from the MEAN
- (x-mean) / std.deviation

```python
import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})
df
```

```python
from scipy.stats import zscore
```

```python
df["zscore_study_hours"] = zscore(df["study_hours"])

print(zscore(df["study_hours"]))
df
```

```python
outliers = df[abs(df["zscore_study_hours"]) > 1]
outliers
```

```python
# Attendence Z_score

df["attendence"] = df["attendance"].fillna(df["attendance"].median())   # Handle Missing value

from scipy.stats import zscore

df["zscore_attendence"] = zscore(df["attendence"])
df
```

```python
abs(df["zscore_attendence"]) > 3   # Universal
```

```python
outliers = df[abs(df["zscore_attendence"]) > 1 ]
outliers
```

# IQR - Inter Quartile Range

```python


import pandas as pd

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

df
```

```python
Q1 = df["sales_amount"].quantile(0.25)
Q3 = df["sales_amount"].quantile(0.75)

print(Q1)
print(Q3)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(IQR)
print(lower)
print(upper)

outliers = df[ (df["sales_amount"]<lower) | (df["sales_amount"]>upper) ]
outliers
```

```python
import pandas as pd

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})
```

```python
Q1 = df["sales_amount"].quantile(0.25)
Q3 = df["sales_amount"].quantile(0.75)

IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

outliers = df[ (df["sales_amount"]<lower_fence) | (df["sales_amount"]>upper_fence) ]
outliers
```

```python
# Removal of Outliers - using IQR

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

Q1 = df["sales_amount"].quantile(0.25)
Q3 = df["sales_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[ (df["sales_amount"]>lower) & (df["sales_amount"]<upper) ]
df
```

```python
# Z score method (outlier removal)

from scipy.stats import zscore

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

df["zscore_sales"] = zscore(df["sales_amount"])
df
```

```python
new_df = df [ abs(df["zscore_sales"]<2) ]
```

# TRANSFORMATION 
 If values are real and we dont want to remove the values. so we transform the values hence the influence much smaller than the original data
 - Log Transformation
- Sqrt transformation

```python
df = pd.DataFrame({

    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

import numpy as np

df["log_sales"] = np.log(df["sales_amount"])
df
```

```python
df["sqrt_sales"] = np.sqrt(df["sales_amount"])
df
```

# Winsorization - Capping and Flooring
Instead of Transforming or removing the Outliers 
Replace with IQR boundary value

```python
import pandas as pd

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

Q1 = df["sales_amount"].quantile(0.25)
Q3 = df["sales_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(upper)

# capping
df["sales_amount"] = np.where(df["sales_amount"]>upper, upper, df["sales_amount"])
df
```

```python
# flooring
print(lower)

df["sales_amount"] = np.where(df["sales_amount"]<lower, lower, df["sales_amount"])
df
```
