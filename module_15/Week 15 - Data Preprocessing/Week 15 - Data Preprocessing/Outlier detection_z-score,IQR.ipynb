```python
import pandas as pd
from scipy.stats import zscore          # Z score inside scipy stat library
```

```python
df = pd.DataFrame({"Salary":[30000,32000,31000,29000,35000,500000]})
df
```

```python
df["Z_score"]= zscore(df["Salary"])
df
```

```python
outliers = df[abs(df["Z_score"]) > 2]  # Generally 3 but 2 taken for this short dataset
print(outliers)                        # abs to convert -ve to +ve. 
```

```python

outlierz = df[ abs(df["Z_score"]) > 3 ]
print(outlierz)
```

IQR

```python
import pandas as pd

df = pd.DataFrame({"Salary":[30000,32000,31000,29000,35000,500000]})
df
```

```python
Q1 = df["Salary"].quantile(0.25)
print(Q1)

Q3 = df["Salary"].quantile(0.75)
print(Q3)

IQR = Q3 - Q1
print(IQR)

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Salary"] < lower) | (df["Salary"] > upper)]   # | bitwise operator means Either
outliers                                                       # Series (a column of Boolean values).  True/False get and convert it.
```
