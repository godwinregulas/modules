```python
import pandas as pd
```

```python
df = pd.DataFrame({"Salary":[30000,32000,31000,29000,35000,500000]})
df
```

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR


print(lower)
print(upper)

# Removal of Outlier Data
new_df = df[ (df["Salary"] >= lower) & (df["Salary"] <= upper) ]  # & bitwise AND 

print(new_df)
```

Transformation

```python
# Log Transformation
import numpy as np

df["Log_Salary"] = np.log(df["Salary"])        # log n (Natural log -> Base e)
df
```

```python
# Square Root Transformation
import numpy as np

df["Sqrt_Salary"] = np.sqrt(df["Salary"])
df
```

```python
# Capping and Flooring - Capping(replace with upper limit), Flooring(replace with lower limit)

#Capping
import numpy as np

df["Salary"] = np.where( df["Salary"] > upper, upper, df["Salary"] )     # numpy.where(condition, [x, y, ]/), Return elements chosen from x or y depending on condition.
df                                                                       # np.where(condition, value_if_true, value_if_false)
```

```python
# Flooring 

df["Salary"] = np.where( df["Salary"] < lower, lower, df["Salary"])   # here we have no lower outlier values if has then replaced with lower limit
```
