Missing values Handling (Filling)

Imputation Methods - Mean, Median, Mode

Mean Imputation

```python
# Mean imputation -  Mean = (Sum of all values) / Total No.of values .
# Syntax    df.fillna()    df[column].fillna( on which manner) here salary column.mean() 
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df
```

Median Imputation

```python
# median  if even entries then middle 2 values add and divide by 2.
df["Age"] = df["Age"].fillna(df["Age"].median())
df
```

```python
# Mode - Most frequently appeared value
df["Experience"] = df["Experience"].fillna(df["Experience"].mode()[0]) 
df               # [0] added here bcz Mode() returns series and 4 It is actually Series([4.0]), so with only mode() not [0] we are passing an entire Series to fillna(), not a single scalar value.  if no modes then not adding [0] then add with mode series but by looking next Mode series value.So every Null filled with different values.LAst null has nothing next to fill with so it remain as NULL   
```

Fill missing values with "Unknown"  - used if the columns are not important to Handle

```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])  # here filled with mode of Gender Text values
df
```

```python
df["Gender"] = df["Gender"].fillna("Unknown")
df
```

UPDATING

```python
# Condition based updation                      df.loc  location
df.loc[df["Age"] == 25, "Experience"] = 2.5     # Syntax df.loc
df
```

```python
# Replace value
df["Gender"]= df["Gender"].replace("Unknown","Known")
df
```
