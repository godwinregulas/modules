Label Encoder

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.DataFrame({"Color":["Red","Green","Blue","Green"]})
df
```

```python
encoder = LabelEncoder()

df["encoded color val"]= encoder.fit_transform(df["Color"])
df
```

One-Hot Encoding

```python
import pandas as pd

df = pd.DataFrame({"Color":["Red","Green","Blue","Green"]})

encoded = pd.get_dummies(df, columns=["Color"])   # so create columns with the name of Color
encoded
```
