```python
import pandas as pd
from sklearn.impute import KNNImputer
```

```python
df = pd.DataFrame({
    "Age":[22,24,23,None],
    "Experience":[1,2,2,2],
    "Salary":[30000,32000,31000,31500]
})
df
```

```python
imputer = KNNImputer(n_neighbors=2)    # imputer set with n neighbour val 2  so calculate within 2 neighbors

new_df = imputer.fit_transform(df)  # (neighbor_value1 + neigh_val2..)/no_of_neigh  --> (24+23)/2 = 23.5
new_df      # 23.5  which is shown here  2.35e+01  bcz of pandas dataframe not numpy operation of fit_transform
```

MICE imputer

```python
# MICE imputer (Multiple Imputation by Chained Equations)   also called Iterative Imputer

import pandas as pd
```

```python
df = pd.DataFrame({"Age":[22,25,30,None],
             "Experience":[1,3,8,5],
             "Salary":[30000,38000,55000,45000]})
df
```

```python
# from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
imputer = IterativeImputer() # random_state=42 inside not mandatory # random_state ensures you get the same result every time  42 commonly used seed value

new_df = imputer.fit_transform(df)
new_df   # 26.93
```
