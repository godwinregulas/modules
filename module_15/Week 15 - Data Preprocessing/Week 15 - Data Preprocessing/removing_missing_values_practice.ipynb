```python
import pandas as pd
```

```python
df = pd.read_csv("employee_missing_values_dataset.csv")
df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df.dropna()      # drop all rows with NULL values
```

```python
df.dropna(axis = 1)  # drop columns with NULL values
```

```python
df.dropna( how="all" )   # drop if all values are  Null in a row
```

```python
df.dropna(thresh = 3)        # min 3 missing values in a row drop
```

```python
df.dropna(subset=["Age"])    # from the given subset column field Null value row deleted
```
