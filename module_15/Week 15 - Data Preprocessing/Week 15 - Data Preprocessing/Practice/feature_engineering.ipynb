Creating new features from Existing Data

```python
# QUANTITY * PRICE = TOTAL    -- new feature
```

```python
import pandas as pd

df = pd.DataFrame({
    "Quantity":[2,5],
    "Price":[500,200]
})

df
```

```python
df["Total"] = df["Quantity"] * df["Price"]
df
```

```python
# Age From DOB

df = pd.DataFrame({
    "DOB":["2003-07-06", "1999-08-15"]
})

df["DOB"] = pd.to_datetime( df["DOB"] )   # from text to real date 

current_year = 2026

df["Age"] = current_year - df["DOB"].dt.year
df
```

Transformation - Log , Polynomial

```python
import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

import numpy as np
df["log_study_hours"] = np.log(df["study_hours"])

df
```

```python
# Polynomial Features -  If Relationships are not linear

from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(include_bias=False)

new = poly.fit_transform(df[["study_hours"]])


df[["study_hour","Polynomial"]] = pd.DataFrame(new, columns=["study_hour", "Polynomial"])
df
```

ENCODING CATEGORICAL VARIABLES - Label, One-hot encoding

```python
# Label Encoding - Each variable get a unique integer
import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["Label_enc_city"] = encoder.fit_transform(df["city"])
df
```

```python
# one-hot encoding - create unique binary table

import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

encoder = OneHotEncoder(sparse_output=False)

encoded = encoder.fit_transform(df[["city"]])

new = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(["city"]))
new


# newq = pd.DataFrame(encoded, columns=["Calicut","Kannur","Kochi","Malappuram","Thrissur"]) #  alphabet order
# newq
```
