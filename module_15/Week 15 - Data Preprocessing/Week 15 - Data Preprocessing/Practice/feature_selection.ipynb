```python
# correlation methods - Measures how strongly each features is related to the Target. Compute Pearson correlation coefficief b/w every pair of numeric columns

import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

numeric_df = df.select_dtypes("number")

target = "study_hours"

corr = numeric_df.corr()
corr[target]
```

```python
# Variance threshold - Features with less variance or no change in values doesn't provide info or contribute to model training so remove it.

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 3, 2, 2, 3, 3, 4, 6],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

from sklearn.feature_selection import VarianceThreshold

selector = VarianceThreshold(threshold=6)

numeric_df = df.select_dtypes("number")

variace_threshold_df = selector.fit_transform(numeric_df)
df = pd.DataFrame(variace_threshold_df, columns=selector.get_feature_names_out())
df
```

```python
# Wrapper Method - RFE, Stepwise Selector

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, 22, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, 45, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, 88],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

target = [5, 3, 2, 2, 3, 3, 4, 6]  # study hr

from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

model = LinearRegression()

selector = RFE(model, n_features_to_select=2)

numerical_df = df.select_dtypes("number")


new_df = selector.fit_transform(numerical_df, target)
df = pd.DataFrame(new_df, columns=selector.get_feature_names_out())


print(selector.ranking_)
print(selector.support_)

df
```

```python
# Stepwise selector - forward selector --> # Start with no feature, Automatically add best features of high R^2 ed features add to it
                    # Backward selector --> Start with all features, Remove least useful features by removing one feature at a time and check  R^2 ed value.

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, 22, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, 45, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, 88],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

numerical_df= df.select_dtypes("number")

target = [5, 3, 2, 2, 3, 3, 4, 6]  # study hr

from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LinearRegression

model = LinearRegression()

selector = SequentialFeatureSelector(model, n_features_to_select=2, direction='forward')

new_df = selector.fit_transform(numerical_df, target)

df = pd.DataFrame(new_df, columns=selector.get_feature_names_out())
df
```

```python
# Lasso , Ridge

# Lasso - Lasso adds a penalty that can make some features coeffiicient exactly zero.

import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, 22, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, 45, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, 88],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

numerical_df = df.select_dtypes("number")
target = [5, 3, 2, 2, 3, 3, 4, 6]  # study hr

from sklearn.linear_model import Lasso

model = Lasso(alpha = 0.1)

new_df = model.fit(numerical_df, target)

print(model.feature_names_in_)
print(model.coef_)

```
