```python
# identifying outliers using statistical methods

# 1. Z-Score  - Z score measures how many standard deviations a value is away from the MEAN

# formula:      (x - Mean) / std.deviation
```

```python
import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

from scipy.stats import zscore

df["Z_score_study_hours"] = zscore(df["study_hours"])
df
```

```python
# ouliers = df["Z_score_study_hours"] > 3
# print(ouliers)

outliers = df[abs(df["Z_score_study_hours"]) > 1]
print(outliers)
```

```python
# FULL single

from scipy.stats import zscore

df["attendance"] = df["attendance"].fillna(df["attendance"].mean())  # had null value either mean or standard deviation become NULL then full become NULL due to Z-Score Formula.

df["Z_score_attendance"] = zscore(df["attendance"])
print(df["Z_score_attendance"])

outliers = df[ abs(df["Z_score_attendance"])>3 ]
ouliers
```

```python
# IQR 

import pandas as pd

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})
```

```python
Q1 = df["sales_amount"].quantile(0.25)
Q3 = df["sales_amount"].quantile(0.75)
print(Q1)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR   # lower limit
print(lower)

upper = Q3 + 1.5 * IQR
print(upper)

outliers = df[ (df["sales_amount"]<lower) | (df["sales_amount"]>upper) ]
outliers
```

HANDLING OUTLIERS

```python
# Removal of Outiers - IQR

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

Q1 = df["sales_amount"].quantile(0.25)
Q3 = df["sales_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[ (df["sales_amount"]>lower) & (df["sales_amount"]<upper)]   # using & (logical and)
df
```

```python
# Z-Score

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

from scipy.stats import zscore

df["zscore_sales_amount"] = zscore(df["sales_amount"])
df = df[ abs(df["zscore_sales_amount"])< 2.5 ]
df
```

TRANSFORMATION

```python
# log Transformation - Natural Logarithm

import numpy as np

df = pd.DataFrame({

    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})


df["Log_sales_amount"] = np.log(df["sales_amount"])
df
```

```python
# sqrt
df["sales_amount"] = np.sqrt(df["sales_amount"])  # Assigned to same feature of sales_amount
df
```

```python
# Capping and flooring - instead of removing or transforming Outliers. Replace with Boundary value.

import pandas as pd

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

Q1 = df["sales_amount"].quantile(0.25)
Q3 = df["sales_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(df)

df["sales_amount"] = np.where(df["sales_amount"]>upper, upper, df["sales_amount"]) # np.where(condition, value_if_true, value_if_false)

df
```

```python
# Flooring

import pandas as pd

df = pd.DataFrame({
    "product": ["Keyboard", "Mouse", "Monitor", "Laptop", "Headset", "Tablet", "Printer", "Camera"],
    "sales_amount": [12000, 7000, 15000, 20000, 9500, 11000, 14000, 250000]
})

Q1 = df["sales_amount"].quantile(0.25)
Q3 = df["sales_amount"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(df)

df["sales_amount"] = np.where(df["sales_amount"]<lower,lower,df["sales_amount"])
df
```
