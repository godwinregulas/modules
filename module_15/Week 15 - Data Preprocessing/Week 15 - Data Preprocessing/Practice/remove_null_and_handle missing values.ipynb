```python
import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

df
```

IMPUTATION METHOD

```python
# math - mean
# age - median
# depat - mode
# att - median

df["maths"] = df["maths"].fillna(df["maths"].mean())

df["age"] = df["age"].fillna(df["age"].median())

df["department"] = df["department"].fillna(df["department"].mode()[0])

df["attendance"] = df["attendance"].fillna(df["attendance"].median())

df
```

ADVANCED IMPUTATION - KNN, MICE

```python
#KNN

import pandas as pd

df = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": ["Afsal", "Anu", "Rahul", "Meera", "Arjun", "Sneha", "Vishnu", "Fathima"],
    "age": [21, 20, None, 21, 23, 20, 220, 21],
    "department": ["Data Science", "Data Science", "Python", "Python", None, "AI", "AI", "Data Science"],
    "maths": [85, 90, 65, 78, None, 92, 74, 88],
    "science": [78, 88, 70, 82, 60, 94, 76, 84],
    "attendance": [92, 95, 75, 88, 65, 96, 80, None],
    "study_hours": [5, 6, 3, 4, 2, 7, 4, 5],
    "city": ["Calicut", "Kochi", "Thrissur", "Kannur", "Calicut", "Kochi", "Thrissur", "Malappuram"]
})

df
```

```python
# for full numeric dataset does need only these lines
from sklearn.impute import KNNImputer  

imputer = KNNImputer(n_neighbors=2)       # if we do with categorical datas(text) the function will try to perform with text that will make error 

df = imputer.fit_transform(df)           

# so we have to take outside the numerical values  and perform KNN on it and merge the values back
```

```python
numeric_df = df.select_dtypes('number')
print(numeric_df)
print("\n")


from sklearn.impute import KNNImputer

imputer = KNNImputer()

imputed_numeric_df = imputer.fit_transform(numeric_df)  # array so we need to transform into DataFrame
print(imputed_numeric_df)
print("\n")


numeric_df = pd.DataFrame(imputed_numeric_df, columns=numeric_df.columns)
print( numeric_df )

df[numeric_df.columns] = numeric_df
df
```

```python
# MICE - Multiple Imputation by Chained Equation

numeric_df = df.select_dtypes("number")

from sklearn.impute import IterativeImputer

imputer = IterativeImputer()

imputed_numeric_df = imputer.fit_transform(numeric_df)

numeric_df = pd.DataFrame(imputed_numeric_df, columns= numeric_df.columns)

df[numeric_df.columns] = numeric_df
df
```
