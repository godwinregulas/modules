```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Arshak\Downloads\employee_missing_values_dataset.csv")      #if the dataset outside the folder then shoud use "r" for reading the file path
df
```

```python
df.isnull()             # check missing values
```

Method 1: Remove Rows with missing values

```python
row_removed_df = df.dropna()    # Remove Rows with missing values
row_removed_df
```

```python
print( df.shape )        # before deletion
row_removed_df.shape     # after deleted
```

Method 2: Remove Columns with Missing value

```python
column_removed_df = df.dropna(axis=1)   # axis 0 - row, axis 1 - column
column_removed_df
```

Other methods : For understanding

Method 3: Remove Rows Only If All Values Are Missing

```python
new = df.dropna(how = "all")   # how parameter for in which manner we want to remove Null value
new
```

Method 4: Remove Rows Based on Specific Columns

```python
new = df.dropna(subset=["Age"])  # if age feature is very important then Only rows with missing Age are removed.
new        
```

Method 5: Keep Rows That Have At Least 3 Non-Missing Values

```python
new = df.dropna(thresh = 3)   # thresh=3 means keep rows that have at least 3 non-missing values.
new
```
