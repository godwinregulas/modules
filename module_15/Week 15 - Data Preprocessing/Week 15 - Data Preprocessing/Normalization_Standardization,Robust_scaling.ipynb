Normalization - MinMax Scailing  $$x_{\text{new}}=\frac{x-\min(x)}{\max(x)-\min(x)}$$

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.DataFrame({"Age":[20,25,30,35,40]})
df
```

```python
scaler = MinMaxScaler()
df["Normalized"] = scaler.fit_transform(df[["Age"]])   # df["Age"] - series(full include metadata), df[["Age"]] list of value of df location age

df
```

Standardization - Z-Score Scailing $$z = \frac{x - \mu}{\sigma}$$


```python
import pandas as pd
from sklearn.preprocessing import  StandardScaler    # Z-score known as Standard Scaler
```

```python
df = pd.DataFrame({"Age":[20,25,30,35,40]})
df
```

```python
scaler = StandardScaler()

df["Standardized"] = scaler.fit_transform(df[["Age"]])
df
```

Robust Scaler  $$x_{\text{new}} = \frac{x - \text{Median}}{\text{IQR}}$$

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler
```

```python
df = pd.DataFrame({"Salary":[30000,32000,34000,35000,500000]})
df
```

```python
scaler = RobustScaler()

df["Robust"] = scaler.fit_transform(df[["Salary"]])
df
```
