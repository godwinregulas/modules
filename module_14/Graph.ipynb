# BFS Python Implementation

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    visited.add(start)

    while queue:
        node = queue.popleft()
        print(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
```

# DFS Using Recursion

```python
def dfs(graph, node, visited):
    if node in visited:
        return

    visited.add(node)
    print(node)

    for neighbor in graph[node]:
        dfs(graph, neighbor, visited)
```

# DFS Implementation

```python
def dfs(graph, start):
    visited = set()
    stack = [start]

    visited.add(start)

    while stack:
        node = stack.pop()      # Pops from the right (LIFO - Stack behavior)
        print(node)

        # Reversing the neighbors is optional, but it keeps the 
        # left-to-right exploration order identical to BFS
        for neighbor in reversed(graph[node]):
            if neighbor not in visited:
                visited.add(neighbor)
                stack.append(neighbor)
```
