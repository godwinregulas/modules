# Bubble Sort

```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

arr = [7,5,3,6,4,8]
print(bubble_sort(arr))
```

Optimized

```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

arr = [7,5,3,6,4,8]
print(bubble_sort(arr))
```

# Insertion Sort

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i-1
        while j>=0 and key < arr[j]:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr

arr = [7, 3, 8, 2, 6, 4, 5]
print(insertion_sort(arr))
```

# Selection Sort

```python
def selection_sort(arr):
    for i in range(len(arr)):
        min = arr[i]
        ind = i
        for j in range(i+1,len(arr)):
            if arr[j] < min:
                min = arr[j]
                ind = j
        arr[ind] = arr[i]
        arr[i] = min
    
    return arr

arr = [6, 3, 8, 2, 7, 1, 5, 4]
print(selection_sort(arr))
```
