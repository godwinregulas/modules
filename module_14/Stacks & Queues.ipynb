# Stack

```python
stack = []

def push(value):
    stack.append(value)
    print(value, "pushed into stack")


def pop():
    if len(stack) == 0:
        print("Stack Underflow")
    else:
        value = stack.pop()
        print(value, "popped from stack")


def display():
    if len(stack) == 0:
        print("Stack is empty")
    else:
        print("Stack elements:")
        for value in reversed(stack):
            print(value)


push(10)
push(20)
push(30)

display()

pop()

display()
```

# Queues

```python
from collections import deque

queue = deque()


def enqueue(value):
    queue.append(value)
    print(value, "enqueued")


def dequeue():
    if not queue:
        print("Queue Underflow")
    else:
        value = queue.popleft()
        print(value, "dequeued")


def display():
    if not queue:
        print("Queue is empty")
    else:
        print("Queue:", list(queue))


enqueue(10)
enqueue(20)
enqueue(30)

display()

dequeue()

display()
```

```python
def is_balanced(s):
    stack = []

    pairs = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for char in s:
        if char in "([{":
            stack.append(char)
        elif char in ")]}":
            if not stack:
                return False
            if stack[-1] != pairs[char]:
                return False

            stack.pop()
    return len(stack) == 0
```
