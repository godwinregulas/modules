```python
#Reverse an array
arr = [1, 2, 3, 4, 5]
l = 0
r = len(arr) - 1

while l < r:
    arr[l],arr[r] = arr[r],arr[l]
    l += 1
    r -= 1
arr

# Time = O(n)
# Space Complexity = O(1)
# Modifies original
# If question says “do it in-place, we must use two-pointer method.
```

```python
#Rotate an array k times
arr = [1, 2, 3, 4, 5]
k = 2

n = len(arr)
k = k % n
rev = arr[n-k:] + arr[:n-k]
print(rev)

# After 5 rotations array comes back to original position.If k = 7,first 5 rotations do nothing.Rotate 7 times → same as rotate 2 times.
# That's why we are taking k%n.
# arr = [A B C D E]
# k = 2
# Last 2 elements = [D E]
# Remaining = [A B C]
# Result = [D E] + [A B C] = [D E A B C]
```

```python
#Get max length subarray with sum 0
arr = [1,  2, -3,  3]

sum = 0
seen = {}
max_length = 0

for i,num in enumerate(arr):
    sum += num

    if sum == 0:
        max_length = i + 1

    if sum not in seen:
        seen[sum] = i
    else:
        length = i - seen[sum]
        max_length = max(length,max_length)

print(max_length)
```

```python
#Maximum Sum Subarray (Kadane’s Algorithm)
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = curr_sum = arr[0]

for num in arr[1:]:
    curr_sum = max(num, curr_sum + num)
    max_sum = max(max_sum, curr_sum)

max_sum
```

```python
#Longest Substring Without Repeating Characters.
s = "xyzzaz"
max_len = 0

for i in range(len(s)):
    res = ""
    for j in range(i,len(s)):
        res += s[j]
        if len(res) != len(set(res)):
            break
        max_len = max(max_len,len(res))
max_len
```

```python
#Longest Substring Without Repeating Characters using sliding window
s = "xyzzaz"
left = 0
seen = set()
max_len = 0

for right in range(len(s)):
    while s[right] in seen:
        seen.remove(s[left])
        left += 1
    seen.add(s[right])
    max_len = max(max_len, right - left + 1)

max_len
```

```python
#Maximum sum of any subarray of size 3 using Sliding Window
arr = [2, 1, 5, 1, 3, 2]
k = 3

window_sum = sum(arr[:k])
max_sum = window_sum

for i in range(k, len(arr)):
    window_sum = window_sum - arr[i - k] + arr[i]
    max_sum = max(max_sum, window_sum)

print(max_sum)
```
