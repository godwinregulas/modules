```python
class node:
    def __init__(self, data):
        self.data = data
        self.next = None

class linkedlist:
    def __init__(self):
        self.head = None
    def append(self, data):
        new = node(data)
        if self.head is None:
            self.head = new
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new
    def display(self):
        temp = self.head
        while temp:
            print(temp.data)
            temp = temp.next
        print('None')


ll = linkedlist()
ll.append(10)
ll.append(20)
ll.append(30)
ll.display()
```

```python
class dnode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None
    
class doublell:
    def __init__(self):
        self.head = None
    def append(self, data):
        new = dnode(data)
        if self.head is None:
            self.head = new
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new
        temp.prev = temp
    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=' -> ')
            temp = temp.next

ll = doublell()
ll.append(10)
ll.append(20)
ll.append(30)
ll.append(40)
ll.append(50)
ll.display()
```

```python
arr = [10, 20, 30, 40, 100]

ll = doublell()
for i in arr:
    ll.append(i)
ll.display()
```

```python
def insert_beginning(self, data):
    new = dnode(data)
    new.next = self.head
    self.head = new
```

```python
def insert_end(self, data):
    new = dnode(data)
    if self.head is None:
        self.head = new
        return
    temp = self.head
    while temp.next:
        temp = temp.next
    temp.next = new
```

```python
def delete(self, value):
    if self.head is None:
        return
    if self.head.data == value:
        self.head = self.head.next
        return
    prev = None
    curr = self.head
    while curr and curr.data != value:
        prev = curr
        curr = curr.next
    if curr:
        prev.next = curr.next
```

```python
def fact(num):
    if num>=0:
        if num == 0:
            return 1
        return num * (fact(num - 1))

print(fact(-1))
```

```python
def fibo(num):
    if num == 0:
        return 0
    if num == 1:
        return 1
    return (fibo(num - 1) + (fibo(num - 2)))

print(fibo(4))
```

```python
def power(x, n):
    if n==0:
        return 1
    return x * power(x, n-1)

power(3, 2)
```

```python
def su(num):
    if num == 0:
        return 0
    return (num % 10) + su(num // 10)

print(su(12345))
```

```python
def rev(s):
    if len(s) == 0:
        return ""
    return rev(s[1:]) + s[0]

print(rev('hello'))
```

```python
def palindrome(s, left, right):
    if left >= right:
        return True
    if s[left] != s[right]:
        return False

    return palindrome(s, left + 1, right - 1)

s = "racecar"
print(palindrome(s, 0, len(s) - 1))
```

```python
def find_max(arr, i):
    if i == len(arr) - 1:
        return arr[i]

    remaining_max = find_max(arr, i + 1)

    if arr[i] > remaining_max:
        return arr[i]
    else:
        return remaining_max

arr = [4, 6, 2, 7, 3]
print(find_max(arr, 0))
```

```python
def add_even(arr, i):
    if i == len(arr) - 1:
        return 0
    if arr[i] % 2 == 0:
        return (arr[i] + add_even(arr, i + 1))
    else:
        return add_even(arr, i + 1)

arr = [4, 6, 2, 7, 3]
print(add_even(arr, 0))
```

```python
def linear_search(arr, i, targ):
    if i == len(arr):
        return "Target not found"
    if arr[i] == targ:
        return "Target found at index "+str(i)
    else:
        return linear_search(arr, i+1, targ)

arr = [4, 6, 2, 7, 3]
print(linear_search(arr, 0, 2))
```

```python
def binary_search(arr, high, low, targ):
    if low>high:
        return "not found"
    mid = (low + high)//2
    if arr[mid] == targ:
        return mid
    elif targ < arr[mid]:
        return binary_search(arr, mid - 1, low, targ)
    else:
        return binary_search(arr, high, mid + 1, targ)

arr = [2,4,5,7,9,11,33,55,78,90]
print(binary_search(arr, len(arr)-1, 0, 55))
```
